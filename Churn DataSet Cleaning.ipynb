```python
# Remove Duplicates
df.drop_duplicates(inplace = True)
```

```python
# Replace Dirty Values
df.replace(['N/A', 'NULL', '', ' '], np.nan, inplace  = True)
```

```python
# Step 7 Remove Extra Spaces
df = df.map(lambda x: x.strip() if isinstance(x, str) else x)
```

```python
#Value Count
df["plan_type"].value_counts()
```

```python
# Step 8 Proper Case
df["plan_type"] = df["plan_type"].str.title()
```

```python
# step 9  standardize churn column if until your dataset contain churn column
df['churn'].value_counts()
```

```python
#Step 10 Convert Numeric Column( if the DataSet Contain numberic value but describe as object)
#cols = ["Age", "Monthly_charge"]

#for c in cols:
    #df[c] = pd.to_numeric(df[c],error = "coerce") # coerce mean set as null
```

```python
#df.describe()
```

```python
# Removing invalid Age (If Dataset Contain invaild Age)
df = df[(df ['Age'] >= 18) & (df ['Age'] < 100)]
```

```python
# Step 11 Removing Negative Value
df = df[df['monthly_fee'] > 0]
```

```python
# checking the shape of the Data After Removing -ve values
df.shape
```

```python
# Step 12 Change The Date Format

df['signup_date']=pd.to_datetime(df['signup_date'],errors='coerce')
```

```python
# Step 13 Filling Missing Cells
df["Age"]=df["Age"].fillna(df["Age"].mean())
df["monthly_fee"]=df["monthly_fee"].fillna(df["monthly_fee"].mean())
df["tenure_months"]=df["tenure_months"].fillna(df["tenure_months"].mean())
```

```python
# Step 14 Feature Enggineering (Creation of New Columns)
# Customer Lifetime Value
df['Customer_Value']=df["monthly_fee"]*df["tenure_months"]
```

```python
df.head()
```

```python
# Tenure Groups "0-12" "13-24" "25-48" "49-72"
bins =[0,12,24,48,72]
label = labels=["0-12 months", "13-24 months", "25-48 months", "49-72 months"]
df['tenure_months']=pd.cut(df['tenure_months'], bins = bins, labels = label)
```

```python
df.head()
```

```python
# Senior Citizen Flags
df["Senior_Flag"]=np.where(df["Age"] >= 60, "Senior","Adult")
```

```python
df.head()
```

```python
# Churn Flags
df["Churn_Flag"]=df["churn"].map({"Yes": 1, "No": 0})
```
